## WEEK 02 HANDOFF

### 1. Point Process & Likelihood formulation
- Successfully implemented the inhomogeneous Poisson process log-likelihood: $\log L = \sum_i \log \lambda(s_i) - \int \lambda(s) ds$ evaluated via second-order trapezoidal quadrature over a fixed 4001-node mesh.
- Verified that the integral term acts as a critical regularizer, preventing rate parameters from running away to infinity.

### 2. Parameterization & Fitter Verification
- Used an unconstrained parameterization ($a$, $\theta$, $k$, $s_0$) to keep physical constraints satisfied naturally inside unconstrained optimization spaces.
- Verified the optimizer with a simulated dataset built via **thinning**; parameters were accurately recovered within replicate-to-replicate statistical spreads.
- Exposed two fitter failure modes: dropping the integral term (loud failure; infinite count runaway) and changing the integration limits (quiet failure; incorrect peak and width determination).

### 3. Time-Rescaling & Goodness of Fit
- Applied the **Time-Rescaling Theorem** using the cumulative intensity function (compensator $\Lambda(s)$) to map event arrivals to a stationary Poisson process of unit rate.
- Resolved catalog day-quantization artifacts by uniform daily jittering, yielding a model that passed the Kolmogorov-Smirnov test at a 5% significance level in 17 of 18 development hemicycles.

### 4. Parameter Identifiability & Physical Plausibility
- Profiled the likelihood of $s_{onset}$ to reveal a strong parameter degeneracy/ridge between $s_0$, $k$, and $\theta$. This degeneracy mathematically limits how well a single hemicycle can constrain the rising edge independently, cementing the case for **parameter pooling** in Week 3.
- Discovered a strong correlation of **+0.941** between the total number of events and the independently measured area amplitude $A$, proving that the sunspot bell curve is a physical manifestation of Poisson emergence rates.
